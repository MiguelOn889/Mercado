# Normalización de datos

```python

# Definir una lista para almacenar los DataFrames
dfs = []

# Diccionario para mapear nombres de meses en español a números
meses = {
    "enero": 1,
    "febrero": 2,
    "marzo": 3,
    "abril": 4,
    "mayo": 5,
    "junio": 6,
    "julio": 7,
    "agosto": 8,
    "septiembre": 9,
    "octubre": 10,
    "noviembre": 11,
    "diciembre": 12
}

# Iterar sobre los archivos CSV, cargar cada uno en un DataFrame y agregar columnas adicionales

for file in csv_files:
    df_temp = pd.read_csv(file)
    
    tienda_name = file.stem # Obtener el nombre del archivo sin la extensión
    partes = tienda_name.split("_") # Dividir el nombre del archivo en partes
    
    tienda = partes[0] # Nombre de la tienda
    dia = int(partes[1]) # Día como entero
    mes_nombre = partes[2].lower() # Nombre del mes en minúsculas
    
    mes_num = meses[mes_nombre] # Convertir el nombre del mes a número usando el diccionario
    año = 2026

    # Crear fecha real
    fecha_completa = pd.Timestamp(year=año, month=mes_num, day=dia)

    # Agregar columnas
    df_temp["tienda"] = tienda
    df_temp["fecha"] = fecha_completa
    df_temp["mes"] = mes_num
    df_temp["año"] = año

    dfs.append(df_temp)
    print(df_temp.columns)  # Verificar que la nueva columna se haya agregado correctamente
    print("------")

```

### normalizar nombres de columnas

```python
# normalizar nombres de columnas para que sean iguales en todos los DataFrames 
for df in dfs:
    df.rename(columns={
        "Precio": "Valor_Unitario",
        "Descripcion": "Producto",
    }, inplace=True)

dfs[6].head()
```

```python
# Eliminar columnas innecesarias "Ref" y "Item" si existen en los DataFrames
for df in dfs:
    if "Ref" in df.columns:
        df.drop(columns=["Ref"], inplace=True)
    if "Item" in df.columns:
        df.drop(columns=["Item"], inplace=True)

for df in dfs:
    print(df.columns)  # Verificar que las columnas se hayan eliminado correctamente
    print("------")
```

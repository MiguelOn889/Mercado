# Normalización de datos

```python

# Definir una lista para almacenar los DataFrames
dfs = []

# Diccionario para mapear nombres de meses en español a números
meses = {
    "enero": 1,
    "febrero": 2,
    "marzo": 3,
    "abril": 4,
    "mayo": 5,
    "junio": 6,
    "julio": 7,
    "agosto": 8,
    "septiembre": 9,
    "octubre": 10,
    "noviembre": 11,
    "diciembre": 12
}

# Iterar sobre los archivos CSV, cargar cada uno en un DataFrame y agregar columnas adicionales

for file in csv_files:
    df_temp = pd.read_csv(file)
    
    tienda_name = file.stem # Obtener el nombre del archivo sin la extensión
    partes = tienda_name.split("_") # Dividir el nombre del archivo en partes
    
    tienda = partes[0] # Nombre de la tienda
    dia = int(partes[1]) # Día como entero
    mes_nombre = partes[2].lower() # Nombre del mes en minúsculas
    
    mes_num = meses[mes_nombre] # Convertir el nombre del mes a número usando el diccionario
    año = 2026

    # Crear fecha real
    fecha_completa = pd.Timestamp(year=año, month=mes_num, day=dia)

    # Agregar columnas
    df_temp["tienda"] = tienda
    df_temp["fecha"] = fecha_completa
    df_temp["mes"] = mes_num
    df_temp["año"] = año

    dfs.append(df_temp)
    print(df_temp.columns)  # Verificar que la nueva columna se haya agregado correctamente
    print("------")

```

```python
dfs[7].head()
```

### normalizar nombres de columnas

```python
# normalizar nombres de columnas para que sean iguales en todos los DataFrames 
for df in dfs:
    df.rename(columns={
        "Precio": "Valor_Unitario",
        "Descripcion": "Producto",
    }, inplace=True)

dfs[6].head()
```

```python
# Eliminar columnas innecesarias "Ref" y "Item" si existen en los DataFrames
for df in dfs:
    if "Ref" in df.columns:
        df.drop(columns=["Ref"], inplace=True)
    if "Item" in df.columns:
        df.drop(columns=["Item"], inplace=True)

for df in dfs:
    print(df.columns)  # Verificar que las columnas se hayan eliminado correctamente
    print("------")
```

```python
# mostrar el indice de los dataframes que no tienen la columna total
for i, df in enumerate(dfs):
    if "Total" not in df.columns:
        print(f"DataFrame en índice {i} no tiene la columna 'Total'")
```

```python
dfs[0].head()
```

```python
# cambiar el nombre de la columna "valor_unitario" a "Total" en el DataFrame que no tiene la columna "Total"
for df in dfs:
    if "Total" not in df.columns and "Valor_Unitario" in df.columns:
        df.rename(columns={"Valor_Unitario": "Total"}, inplace=True)

dfs[0].head()
```

```python

# agregar la columna "valor_unitario" a los DataFrames que no tienen la columna "Valor_Unitario" y calcular su valor dividiendo la columna "Total" entre la cantidad de productos
for df in dfs:
    if "Valor_Unitario" not in df.columns and "Total" in df.columns and "Cantidad" in df.columns:
        df["Valor_Unitario"] = df["Total"] / df["Cantidad"] 

dfs[0].head()
```

```python
for df in dfs:
    print(df.columns)  # Verificar que la nueva columna se haya agregado correctamente
    print("------")
```

```python
# ordenar las columnas de cada DataFrame en el mismo orden
column_order = ["Producto", "Cantidad", "Unidad", "Valor_Unitario", "Total", "tienda", "fecha", "mes", "año"]

dfs = [df[column_order] for df in dfs]


for df in dfs:
    print(df.columns)  # Verificar que las columnas estén en el mismo orden
    print("------")
```

# CONTATENAR LOS DATAFRAMES

```python
# Concatenar todos los DataFrames en uno solo
df_final = pd.concat(dfs, ignore_index=True)
df_final.head(100)
```

```python
# Filtrar donde la columna "valor_unitario" es igual a "total" y la cantidad es diferente de 1
df_filtrado = df_final[(df_final["Valor_Unitario"] == df_final["Total"]) & (df_final["Cantidad"] != 1)]

# ajustar el valor de la columna "valor_unitario" dividiendo el valor de la columna "total" entre la columna "cantidad" en el DataFrame filtrado
df_final.loc[df_filtrado.index, "Valor_Unitario"] = df_filtrado["Total"] / df_filtrado["Cantidad"]

df_final.head(100)
```

```python
#  convertir lower case la columna "producto" en el DataFrame final
df_final["Producto"] = df_final["Producto"].str.lower()
df_final.head(100)
```

```python
# decimales de valor_unitario y total a 2 decimales
df_final["Valor_Unitario"] = df_final["Valor_Unitario"].round(2)
df_final["Total"] = df_final["Total"].round(2)
df_final.head(100)
```

```python
# Total gastado en cada tienda
total_por_tienda = df_final.groupby("tienda")["Total"].sum().reset_index()
total_por_tienda.rename(columns={"Total": "Total_Gastado"}, inplace=True)
total_por_tienda.head()
```

```python
# total gastado por mes
total_por_mes = df_final.groupby("mes")["Total"].sum().reset_index()
total_por_mes.rename(columns={"Total": "Total_Gastado"}, inplace=True)
total_por_mes.head()
```
